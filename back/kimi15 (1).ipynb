```python
import json
import re
from groq import Groq

client = Groq(api_key=os.getenv("GROQ_API_KEY"))

def extract_relations(segment):
    prompt = f"""
Vous êtes un expert en extraction d'informations juridiques.
Extrayez toutes les entités et leurs relations à partir du texte suivant.

RÈGLES STRICTES :
- Extrayez UNIQUEMENT les entités et relations EXPLICITEMENT présentes dans le texte.
- Chaque triplet doit être UNIQUE (pas de doublons).
- Chaque entité doit avoir un "name" et un "type" précis (personne, organisation, décret, arrêté, institution, fonction, etc.).
- La relation doit être un verbe unique seulement extrait du texte, sans prépositions ni phrases.
- Identifiez le sujet, le verbe et l'objet même si la formulation est complexe. 
- Le texte peut contenir plusieurs paragraphes ou blocs indépendants.
- Chaque bloc doit être traité séparément.
- NE PAS mélanger les informations entre deux paragraphes.
- Les personnes listées sous un paragraphe prennent UNIQUEMENT le rôle mentionné dans CE paragraphe.
EXEMPLE :
Texte : Le président nomme le ministre.
Sortie :
[
{{
  "entity1": {{"name": "président", "type": "personne"}},
  "relation": "nommer",
  "entity2": {{"name": "ministre", "type": "fonction"}}
}}
]

Retournez UNIQUEMENT un JSON valide, sans texte avant ou après.

Text:
{segment}
"""

    completion = client.chat.completions.create(
        model="meta-llama/llama-4-scout-17b-16e-instruct",
        messages=[{"role": "user", "content": prompt}],
        temperature=0.1,
        max_tokens=1024
    )

    response = completion.choices[0].message.content

    print("=== RAW OUTPUT ===")
    print(response)
    print("==================")

    match = re.search(r'\[.*\]', response, re.DOTALL)
    if match:
        try:
            return json.loads(match.group())
        except json.JSONDecodeError as e:
            print(f"JSON parse error: {e}")
            print(f"Tried to parse: {match.group()}")

    return []
```

```python
def clean_all_results(data):

    if isinstance(data, str):
        cleaned = data.replace('\n', ' ').strip()
        cleaned = ' '.join(cleaned.split())
        return cleaned

    elif isinstance(data, list):
        cleaned_list = [clean_all_results(item) for item in data]
        return [item for item in cleaned_list if item]   

    elif isinstance(data, dict):
        return {key: clean_all_results(value) for key, value in data.items()}

    else:
        return data
```

```python
import json
import re

file_path = "/kaggle/input/datasets/cherniwajih/27dec21/kimi27dec2021.json"

with open(file_path, "r", encoding="utf-8") as f:
    raw_data = json.load(f)

if not isinstance(raw_data, list):
    raise TypeError("raw_data n'est pas une liste, impossible de boucler.")

# ✅ Validation functions
def is_valid_entity(entity):
    return (
        isinstance(entity, dict) and
        "name" in entity and
        "type" in entity and
        isinstance(entity["name"], str) and entity["name"].strip() != "" and
        isinstance(entity["type"], str) and entity["type"].strip() != ""
    )

def is_valid_triple(triple):
    return (
        isinstance(triple, dict) and
        "entity1" in triple and
        "entity2" in triple and
        "relation" in triple and
        isinstance(triple["relation"], str) and triple["relation"].strip() != "" and
        is_valid_entity(triple["entity1"]) and
        is_valid_entity(triple["entity2"])
    )

all_relations = []
rejected = []

for idx, text_block in enumerate(raw_data):
    print(text_block)

    # ✅ cas 1 : liste
    if isinstance(text_block, list):
        for triple in text_block:
            if is_valid_triple(triple):
                all_relations.append(triple)
                print(f"✅ Élément {idx} : relation ajoutée (liste)")
            else:
                rejected.append((idx, triple))
                print(f"❌ Élément {idx} rejeté (liste)")
        continue

    # ✅ cas 2 : dict
    if isinstance(text_block, dict):
        if is_valid_triple(text_block):
            all_relations.append(text_block)
            print(f"✅ Élément {idx} : relation ajoutée (dict)")
        else:
            rejected.append((idx, text_block))
            print(f"❌ Élément {idx} rejeté (dict)")
        continue

    # ✅ cas 3 : string (LLM)
    if isinstance(text_block, str):
        match = re.search(r'\[.*\]', text_block, re.DOTALL)

        if match:
            json_string = match.group(0)

            try:
                data = json.loads(json_string)

                if isinstance(data, list):
                    for triple in data:
                        if is_valid_triple(triple):
                            all_relations.append(triple)
                            print(f"✅ Élément {idx} : relation ajoutée (string)")
                        else:
                            rejected.append((idx, triple))
                            print(f"❌ Élément {idx} rejeté (string)")

            except json.JSONDecodeError as e:
                print(f"❌ Élément {idx} : erreur JSON -> {e}")
        continue

    # ❌ autre type
    print(f"⚠️ Élément {idx} ignoré (type: {type(text_block)})")


print("\n✅ Total relations valides :", len(all_relations))
print("❌ Total rejetées :", len(rejected))
```

```python
def normalize_label(label):
    mapping = {
        "personne": "Person",
        "organisation": "Organization",
        "institution": "Organization",
        "fonction": "Function",
        "décret": "Law",
        "arrete": "Law",
        "texte juridique": "Document"
    }
    return mapping.get(label.lower().strip(), "Entity")
```

```python
def normalize_relation(rel):
    rel = rel.strip().upper().replace(" ", "_")
    rel = re.sub(r'[^A-Z0-9_]', '', rel)
    return rel if rel else "RELATED_TO"
```

```python
def escape_relation(rel):
    # remplace les backticks internes par underscore
    rel = rel.replace("`", "_")
    return f"`{rel}`"
```

```python
def triplet_exists(tx, e1_name, relation, e2_name):
    relation_safe = escape_relation(relation)
    query = f"""
    MATCH (a {{name: $e1_name}})-[r:{relation_safe}]->(b {{name: $e2_name}})
    RETURN COUNT(r) > 0 AS exists
    """
    result = tx.run(query, e1_name=e1_name, e2_name=e2_name)
    record = result.single()
    exists = record and record["exists"]
    
    print(f"DEBUG → exists={exists} | e1='{e1_name}' | rel='{relation_safe}' | e2='{e2_name}'")
    
    return exists


def create_graph(tx, e1_name, e1_type, relation, e2_name, e2_type):
    
    if triplet_exists(tx, e1_name, relation, e2_name):
        print(f"Triplet already exists: ({e1_name})-[{relation}]->({e2_name})")
        return False

    relation_safe = escape_relation(relation)
    query = f"""
    MERGE (a:{e1_type.capitalize()} {{name: $e1_name}})
    MERGE (b:{e2_type.capitalize()} {{name: $e2_name}})
    MERGE (a)-[:{relation_safe}]->(b)
    """
    tx.run(query, e1_name=e1_name, e2_name=e2_name)
    return True
```

```python
with driver.session() as session:
    for i, triple in enumerate(all_relations):
        if "entity1" in triple and "entity2" in triple:
            try:
                session.execute_write(
                    create_graph,
                    triple["entity1"]["name"],
                    triple["entity1"]["type"],
                    triple["relation"],
                    triple["entity2"]["name"],
                    triple["entity2"]["type"]
                )
            except Exception as e:
                print(f"❌ Error at triple {i}: {e}")
        else:
            print(f"⚠️ Triple {i} incomplet :", triple)

driver.close()
```
